```python
def sum_of_two(arr:list, target:int):
    set_nums = set()
    result = []
    for num in arr:
        if (target - num) in set_nums:
            result.append((target-num,num))
        else:
            set_nums.add(num)
    
    return result
```

```python
arr = [1,3,5,8,10]
target = 13
sum_of_two(arr,target)
```

```python
arr = [2, 3, 4, 3]
target = 6
sum_of_two(arr,target)
# Ожидаемый результат: [(2,4), (3,3)]
# Твой код выдаст: [(2,4), (3,3), (3,3)] (повтор)
```

```python
# tuple — можно (хешируемый)
d = {(1, 2): "point"}
print(d[(1, 2)])  # ✅

# list — нельзя (не хешируемый)
d = {[1, 2]: "point"}  # ❌ TypeError: unhashable type: 'list'
```

```python
def squares(n):
    result = []
    for i in range(n):
        result.append(i ** 2)
    return result

print(squares(5))  # [0, 1, 4, 9, 16]
```

```python
def squares_gen(n):
    for i in range(n):
        yield i ** 2
```

```python
gen = squares_gen(5)
print(gen)  # <generator object squares_gen at 0x...>
```

```python
for val in gen:
    print(val)  # 0, 1, 4, 9, 16 (по одному)
```

```python
def infinite_count():
    i = 0
    while True:
        yield i
        i += 1

counter = infinite_count()
print(next(counter))  # 0
print(next(counter))  # 1
print(next(counter))  # 2
# ... можно продолжать бесконечно
```

```python
import pandas as pd

df = pd.DataFrame({
    'group': ['A', 'A', 'B', 'B', 'B'],
    'value': [10, 20, 30, 40, 50]
})

# agg — одна строка на группу
agg_result = df.groupby('group')['value'].mean()
print(agg_result)
# group
# A    15.0
# B    40.0
# Name: value, dtype: float64

print(len(agg_result))  # 2
```

```python
df.groupby('group')['value'].transform('mean')
```

```python
# transform — та же длина, что и исходный df
transform_result = df.groupby('group')['value'].transform('mean')
print(transform_result)
# 0    15.0
# 1    15.0
# 2    40.0
# 3    40.0
# 4    40.0
# Name: value, dtype: float64

print(len(transform_result))  # 5 (как исходный df)
```

```python
df['group_mean'] = df.groupby('group')['value'].transform('mean')
print(df)
#   group  value  group_mean
# 0     A     10        15.0
# 1     A     20        15.0
# 2     B     30        40.0
# 3     B     40        40.0
# 4     B     50        40.0
```

```python
df['deviation'] = df['value'] - df.groupby('group')['value'].transform('mean')
print(df)
#   group  value  group_mean  deviation
# 0     A     10        15.0       -5.0
# 1     A     20        15.0        5.0
# 2     B     30        40.0      -10.0
# 3     B     40        40.0        0.0
# 4     B     50        40.0       10.0
```

```python

df = pd.DataFrame({
    'date': ['2024-01-01', '2024-01-02', '2024-01-03'],
    'product': ['A', 'B', 'A'],
    'sales': [100, 200, 150]
})
df
```

```python

pivot_df = df.pivot(index='date', columns='product', values='sales')
pivot_df
# product      A    B
# date
# 2024-01-01 100  NaN
# 2024-01-02 NaN  200
# 2024-01-03 150  NaN
```

```python
df_duplicate = pd.DataFrame({
    'date': ['2024-01-01', '2024-01-01', '2024-01-02'],
    'product': ['A', 'A', 'B'],
    'sales': [100, 200, 150]
})

pivot_df = df_duplicate.pivot(index='date', columns='product', values='sales')
pivot_df
```

```python
# pivot_table автоматически агрегирует дубли (по умолчанию mean)
pivot_table_df = df_duplicate.pivot_table(
    index='date',
    columns='product',
    values='sales',
    aggfunc='mean'  # можно sum, count, first, и т.д.
)
print(pivot_table_df)
# product        A    B
# date
# 2024-01-01  150.0  NaN
# 2024-01-02  NaN   150.0
```
